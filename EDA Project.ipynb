```python
sns.countplot(x = 'Gender', data = df)
```

```python
ax = sns.countplot(x = 'Gender', data = df)
for i in ax.containers : 
    ax.bar_label(i)
```

```python
df.groupby('Gender', as_index = False)['Amount'].sum().sort_values(by = 'Amount', ascending = False)
```

```python
sales_gen = df.groupby('Gender', as_index = False)['Amount'].sum().sort_values(by = 'Amount', ascending = False)
sns.barplot(x = 'Gender', y = 'Amount', data = sales_gen)
```

```python
# Age Group
```

```python
df.columns
```

```python
df.groupby('Age Group', as_index = False)['Amount'].sum().sort_values(by = 'Amount', ascending = False)
```

```python
sns.countplot(x = 'Age Group', hue = 'Gender', data = df)
```

```python
ax = sns.countplot(x = 'Age Group', hue = 'Gender', data = df)
for i in ax.containers:
    ax.bar_label(i)
```

```python
# Age Group vs Amount
axes = df.groupby(['Age Group'], as_index = False)['Amount'].sum().sort_values(by = 'Amount',ascending = False)
sns.barplot(x = 'Age Group', y = 'Amount', data = axes)
```

```python
df.groupby('State', as_index = False)['Orders'].sum().sort_values(by = 'Orders', ascending = False).head(10)
```
